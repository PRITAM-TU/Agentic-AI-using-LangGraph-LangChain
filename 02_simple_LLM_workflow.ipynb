```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from google import genai
import os
load_dotenv()  # Load environment variables from .env file
```

```python
model= ChatGroq(
    model="openai/gpt-oss-120b",
    groq_api_key=os.getenv("GROQ_API_KEY")
)
```

```python
class LLMState(TypedDict):
    input_text: str
    output_text: str
```

```python
#grapg defination 
graph=StateGraph(LLMState)
```

```python
def LLM_call(state: LLMState) -> LLMState:
    # Use the model to generate output based on the input text
    #take the question from the state and pass it to the model
    input_text = state['input_text']
    #call the model and get the response
    response = model.invoke(input_text).content
    #store the response in the state
    state['output_text'] = response
    #return the state
    return state
```

```python
#add the node into the graph

graph.add_node("llm_call", LLM_call)
graph.add_edge(START, "llm_call")
graph.add_edge("llm_call", END) 
workflow=graph.compile()  # Compile the graph to prepare for execution
```

```python
initial_state={"input_text": "What is the capital of France?", "output_text": ""}
final_state=workflow.invoke(initial_state)  # Execute the workflow with the initial state
final_output=final_state['output_text'] 
print(final_output)  # Extract the output text from the final state
```
